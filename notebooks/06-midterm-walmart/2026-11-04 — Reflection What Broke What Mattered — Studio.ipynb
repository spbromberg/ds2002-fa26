### 1. Data quality

Duplicate transaction IDs were the most consequential data-quality issue for the headline demand number. The Q1 pipeline removes 30 repeated transaction IDs. With deduplication, the September 23–25 average is 3,504 units/day against a September 8–15 baseline of 1,091.25 units/day (3.21×). ### 2. ETL trade-offs

Name a cleaning decision where a reasonable person could have chosen the other way. What was the other option, what would it have done to your reported numbers, and would you make the same call again?

SKU consolidation is a judgment call: we mapped 45 raw SKU labels to 20 canonical products. A reasonable alternative would be to keep all 45 labels separate and avoid combining variants unless the identity is clear. That would make product-level counts more fragmented and could hide demand for the same product under different codes; it would not change the deduplicated total of 49,343 units, as the notebook's assertion confirms. I would consolidate again, but keep the mapping explicit and review uncertain aliases before using product-level results.
